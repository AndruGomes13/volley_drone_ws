# Vicon RTS Smooting

This filters the Vicon estimates with an ESEKF and subsequent RTS smoothing. In principle, it should provide much better velocity and body rate estimates.

```python

DEG = np.pi / 180.0  # Degrees to radians conversion factor

# noise_params = eskf.NoiseParams(
#     sigma_a=0.2,           # Process noise on acceleration (m/s^2/√Hz)
#     sigma_alpha=0.1,       # Process noise on ang. acceleration (rad/s^2/√Hz)
#     # Vicon measurement noise parameters (should be tuned based on the static analysis)
#     sigma_p_meas=0.001,    # Vicon position measurement noise (m)
#     sigma_theta_meas=0.1*DEG  # Vicon orientation measurement noise (rad)
# )
noise_params = eskf.NoiseParams(
    sigma_a=0.001,           # Process noise on acceleration (m/s^2/√Hz)
    sigma_alpha=0.1,       # Process noise on ang. acceleration (rad/s^2/√Hz)
    # Vicon measurement noise parameters (should be tuned based on the static analysis)
    sigma_p_meas=0.001,    # Vicon position measurement noise (m)
    sigma_theta_meas=0.01*DEG  # Vicon orientation measurement noise (rad)
)

initial_state = eskf.NominalState(p=vicon_measurements.position.to_numpy()[0], v=np.zeros(3), q=vicon_measurements.orientation_xyzw.to_numpy()[0], w=np.zeros(3))
ekf = eskf.ViconESEKFConstOmega(state=initial_state, noise=noise_params)
ts= vicon_measurements.time.to_numpy()
dts = np.diff(ts)

history: List[eskf.FilterHistoryPoint] = []
for k, dt in enumerate(dts):
        # Predict step
        ekf.predict(dt)
        
        # Store predicted state
        x_pred = eskf.copy_state(ekf.x)
        P_pred = ekf.cov.P.copy()
        F      = ekf.F_discrete(dt)  # OK to recompute here *before* update
        
        meas_p = vicon_measurements.position.to_numpy()[k + 1]
        meas_q = vicon_measurements.orientation_xyzw.to_numpy()[k + 1]

        # Update step
        ekf.update(meas_p, meas_q)
        
        x_upd = eskf.copy_state(ekf.x)
        P_upd = ekf.cov.P.copy()
        
        # Store results for this step
        history.append(eskf.FilterHistoryPoint(
            t=ts[k+1], dt=dt, F=F,
            x_pred=x_pred, P_pred=P_pred,
            x_upd=x_upd, P_upd=P_upd
        ))

smoothed_states , smoothed_timestamps= eskf.rts_smooth(history)

p_filtered = np.array([s.x_pred.p for s in history])
v_filtered = np.array([s.x_pred.v for s in history])
q_filtered = np.array([s.x_pred.q for s in history])
w_filtered = np.array([s.x_pred.w for s in history])
ts_filtered = np.array([s.t for s in history])

p_smth = np.array([s.p for s in smoothed_states])
v_smth = np.array([s.v for s in smoothed_states])
q_smth = np.array([s.q for s in smoothed_states])
w_smth = np.array([s.w for s in smoothed_states])
ts_smth = np.array(smoothed_timestamps)
```

```python
# --- Interpolation ---
p_smth_interp, mask = au.interpolate_xyz_vector(p_smth, ts_smth, state_estimates.time.to_numpy())

R_smth = R.from_quat(q_smth)
R_smth_interp, mask = au.interpolate_orientation(R_smth, ts_smth, state_estimates.time.to_numpy())

v_smth_interp, mask = au.interpolate_xyz_vector(
    v_smth,
    ts_smth,
    state_estimates.time.to_numpy()
)

w_smth_interp, mask = au.interpolate_xyz_vector(
    w_smth,
    ts_smth,
    state_estimates.time.to_numpy()
)

ts_smth_interp = state_estimates.time.to_numpy()[mask]

interp_mask = mask
assert ts_smth_interp.shape[0] == p_smth_interp.shape[0] == v_smth_interp.shape[0] == w_smth_interp.shape[0] == len(R_smth_interp)
```

```python
import numpy as np
from scipy.signal import correlate, correlation_lags
from scipy.interpolate import interp1d

def estimate_delay(x, y, dt):
    # zero-mean
    x0 = x - x.mean()
    y0 = y - y.mean()
    c = correlate(y0, x0, mode="full", method="fft")
    lags = correlation_lags(len(y0), len(x0), mode="full")  # samples
    k_idx = int(np.argmax(c))
    lag_samples = int(lags[k_idx])
    tau = lag_samples * dt  # seconds; tau>0 => y lags x
    return lag_samples, tau, c, k_idx

def refine_subsample_peak(c, k_idx, dt):
    # parabolic interpolation around the discrete peak
    if k_idx <= 0 or k_idx >= len(c) - 1:
        return 0.0  # can't refine at boundary
    y1, y2, y3 = c[k_idx - 1], c[k_idx], c[k_idx + 1]
    denom = (y1 - 2*y2 + y3)
    if denom == 0:
        return 0.0
    delta_samples = 0.5 * (y1 - y3) / denom  # in (-0.5, 0.5)
    return float(delta_samples * dt)  # seconds

def shift_signal(t, s, tau):
    # returns s(t - tau); positive tau delays s
    f = interp1d(t, s, kind="linear", fill_value="extrapolate", assume_sorted=True)
    return f(t - tau)

# ----- usage -----
dt = float(np.diff(ts_smth_interp).mean())  # safer than [0]
ang_vel_interp_body = R_smth_interp.inv().apply(w_smth_interp)
i = 2
x = ang_vel_interp_body[:, i]                    # reference on ts_smth_interp
y = state_estimates.velocity_angular.to_numpy()[interp_mask, i]
t_y = ts_smth_interp                        # make sure y is on the same grid (interpolated)

lag, tau, c, k_idx = estimate_delay(x, y, dt)
delta = refine_subsample_peak(c, k_idx, dt)  # seconds (can be +/-)

# If tau>0, y lags x. To align, ADVANCE y by (tau+delta) => pass negative to shift_signal
print(f"lag={lag}, tau={tau:.5f} s, delta={delta:.5f} s")
y_aligned = shift_signal(t_y, y, -(tau + delta))

lag2, tau2, _, _ = estimate_delay(x, y_aligned, dt)

print(f"Before alignment: { np.rad2deg(np.max(y - x)):.5f}")
print(f"Before alignment: { np.rad2deg(np.max(y_aligned - x)):.5f}")

```
